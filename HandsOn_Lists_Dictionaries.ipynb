## 1.2 Apa itu List?

**List** adalah wadah berurutan yang bisa menyimpan banyak item dalam satu variabel. Ciri-cirinya:

- Ditulis dengan kurung siku `[ ]`
- Urutan item **selalu diingat**

- Diakses lewat nomor **index**, dimulai dari `0`

## 1.3 Menambah & Menghapus Item

List itu fleksibel — isinya bisa berubah kapan saja dengan beberapa method berikut:

| Method | Fungsi |
|---|---|
| `.append(item)` | menambah item di **akhir** list |
| `.insert(index, item)` | menyisipkan item di **posisi tertentu** |
| `.remove(item)` | menghapus item **tertentu** (berdasarkan nilainya) |

````python
nama = ["Ana", "Budi", "Cici"]

# Menambah di akhir
nama.append("Dedi")
print(nama)   # ['Ana', 'Budi', 'Cici', 'Dedi']

# Menyisipkan di posisi tertentu (index 1)
nama.insert(1, "Eni")
print(nama)   # ['Ana', 'Eni', 'Budi', 'Cici', 'Dedi']

# Menghapus item tertentu
nama.remove("Budi")
print(nama)   # ['Ana', 'Eni', 'Cici', 'Dedi']
````

## 1.5 Indexing — Mengakses Item List

Setiap item punya "alamat" berupa nomor index. Index dari depan mulai dari `0`, dan index dari belakang mulai dari `-1`.

````python
nama = ["Ana", "Budi", "Cici", "Dedi", "Eni"]
#          0       1       2       3       4
#         -5      -4      -3      -2      -1

print(nama[0])     # Ana   -> item pertama
print(nama[2])     # Cici
print(nama[-1])    # Eni   -> item terakhir (dari belakang)
````

## 1.6 Slicing — Mengambil Sebagian List

Gunakan `[awal:akhir]` untuk mengambil beberapa item sekaligus. **Ingat**: index akhir **tidak ikut** terambil!

````python
nama = ["Ana", "Budi", "Cici", "Dedi", "Eni"]

print(nama[1:3])     # ['Budi', 'Cici']       -> index 1 s.d. 2
print(nama[:2])      # ['Ana', 'Budi']        -> dari awal sampai index 1
print(nama[::2])     # ['Ana', 'Cici', 'Eni']  -> loncat 2 index sekali ambil
````

````python
nama[::-2]
````

# 📖 Bagian 2: Dictionary

## 2.1 Apa itu Dictionary?

**Dictionary** menyimpan data berpasangan **key → value**, seperti kartu nama: setiap informasi (key) punya nilainya sendiri (value).

- Ditulis dengan kurung kurawal `{ }`
- Setiap pasangan ditulis `"key": value`
- **Key harus unik** (tidak boleh ada 2 key yang sama), tapi value boleh apa saja

````python
profil = {
    "nama": "Budi",
    "umur": 20,
    "kota": "Jakarta"
}

print(profil)
print(type(profil))
````

## 2.2 Mengakses Value lewat Key

Dictionary tidak diakses dengan nomor index, melainkan lewat **nama key**-nya.

````python
profil = {"nama": "Budi", "umur": 20}

print(profil["nama"])          # Budi

# Cara aman, tidak error jika key tidak ada
print(profil.get("kota"))      # None (karena key "kota" belum ada)
````

````python
profil["kota"]
````

> ⚠️ **Hati-hati:** `profil["kota"]` akan menghasilkan `KeyError` jika key `"kota"` tidak ada. Gunakan `.get()` agar lebih aman.

## 2.3 Menambah & Mengubah Dictionary

- Jika key **sudah ada** → value diperbarui
- Jika key **belum ada** → key baru langsung ditambahkan

````python
profil = {"nama": "Budi", "umur": 20}

# Menambah key baru
profil["kota"] = "Jakarta"

# Mengubah value yang sudah ada
profil["umur"] = 21

print(profil)   # {'nama': 'Budi', 'umur': 21, 'kota': 'Jakarta'}
````

✏️ **Coba Sendiri:** Buat dictionary bernama `siswa` berisi `nama`, `kelas`, dan `nilai` milikmu (boleh rekaan). Lalu ubah nilai `nilai` menjadi 10 poin lebih tinggi.

````python
# Tulis kodemu di sini
````

## 2.4 Iterasi Dictionary dengan Loop

Gunakan `.items()` untuk mengambil **key dan value sekaligus** di setiap perulangan.

````python
profil = {"nama": "Budi", "umur": 20, "kota": "Jakarta"}

for key, value in profil.items():
    print(key, "→", value)
````

````python
for key in profil.keys():
  print(key)
````

````python
for value in profil.values():
  print(value)
````

## 2.5 List vs Dictionary — Kapan Pakai yang Mana?

| | **List** | **Dictionary** |
|---|---|---|
| Akses data | lewat index angka `list[0]` | lewat key/nama `dict["kunci"]` |
| Cocok untuk | data berurutan/sejenis | data berlabel/berpasangan |
| Contoh | daftar nilai, daftar belanja | profil siswa, data kontak |
| Sintaks | `nama = [item1, item2, ...]` | `nama = {"key": value}` |

## 2.6 Bonus: Kombinasi List & Dictionary

List dan Dictionary bisa digabung — List berisi banyak Dictionary. Ini sangat umum dipakai untuk menyimpan data banyak "record", misalnya data banyak siswa sekaligus.

````python
siswa = [
    {"nama": "Ana", "nilai": 90},
    {"nama": "Budi", "nilai": 85},
]

for s in siswa:
    print(s["nama"], s["nilai"])
````

# 🔒 Bagian 3 (Bonus): Tuple

**Tuple** mirip List, tapi datanya **terkunci** — tidak bisa diubah (*immutable*) setelah dibuat.

- Ditulis dengan kurung `( )`, bukan `[ ]`
- Cocok untuk data yang **harus tetap**, misalnya koordinat `(x, y)` atau kode warna RGB

````python
koordinat = (10, 20, 30)

print(koordinat[0])       # 10  -> tetap bisa diakses lewat index seperti List
print(len(koordinat))     # 3
````

````python
# Mencoba mengubah isi tuple akan menghasilkan error
koordinat = (10, 20, 30)
koordinat[0] = 99   # ❌ Jalankan sel ini, perhatikan errornya!
````

> 🔎 Jalankan sel di atas — kamu akan melihat `TypeError: 'tuple' object does not support item assignment`. Itu **wajar**, karena memang itulah sifat utama Tuple: datanya tidak bisa diubah.

**List vs Tuple:**
- `List [ ]` → bisa diubah (*mutable*)
- `Tuple ( )` → tidak bisa diubah (*immutable*)

# 🔢 Bagian 5 (Bonus): NumPy Array

Untuk data angka dalam jumlah besar, List biasa terasa lambat. **NumPy** adalah library Python yang menyediakan struktur data **array** yang jauh lebih cepat dan hemat memori untuk komputasi numerik.

> Jika belum terpasang, jalankan `!pip install numpy` terlebih dahulu di sel baru.

````python
import numpy as np

data = np.array([10, 20, 30, 40])

print(data)            # [10 20 30 40]
print(type(data))
print(data.shape)      # (4,)  -> artinya array 1 dimensi dengan 4 elemen
````

**Kelebihan utama NumPy** dibanding List biasa: operasi matematis bisa langsung diterapkan ke **semua elemen sekaligus** (disebut *vectorized operation*), tanpa perlu loop manual.

````python
data = np.array([10, 20, 30, 40])

print(data * 2)     # [20 40 60 80]   -> dikalikan 2 tanpa loop!
print(data + 5)     # [15 25 35 45]
print(data.sum())   # 100
print(data.mean())  # 25.0

# Bandingkan dengan cara List biasa (butuh loop / list comprehension):
data_list = [10, 20, 30, 40]
hasil = [x * 2 for x in data_list]
print(hasil)         # [20, 40, 60, 80]
````

NumPy array juga bisa berbentuk **matriks (2 dimensi)**:

````python
matrix = np.array([[1, 2], [3, 4]])

print(matrix)
print(matrix.shape)   # (2, 2) -> 2 baris, 2 kolom
````

# 🧮 Bagian 6 (Bonus): Operasi NumPy untuk Aljabar Linear

NumPy menyediakan operasi matriks siap pakai lewat modul `numpy.linalg`, yang menjadi dasar penting untuk *machine learning* dan *data science*.

| Operasi | Kode |
|---|---|
| Perkalian matriks | `A @ B` atau `np.dot(A, B)` |
| Transpose (baris ↔ kolom) | `A.T` |
| Determinan | `np.linalg.det(A)` |
| Invers matriks | `np.linalg.inv(A)` |

````python
import numpy as np

A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print("Perkalian matriks (A @ B):")
print(A @ B) #dot product
````

````python
np.cross(A, B) #cross product
````

````python
print("Transpose A (A.T):")
print(A.T)
````

````python
print("Determinan A:")
print(np.linalg.det(A))
````

````python
print("Invers matriks A:")
print(np.linalg.inv(A))
````
