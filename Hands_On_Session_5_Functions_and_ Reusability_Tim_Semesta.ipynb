## 2️⃣ Apa Itu Fungsi?

Fungsi itu seperti **mesin**: kita masukkan sesuatu (**input**), mesin memprosesnya, lalu keluar **hasil** (**output**).

> 🥤 **Analogi Blender**: Masukkan **buah** (input) → Blender **memproses** (fungsi) → Keluar **jus** (output)

Dalam Python, pola yang sama berlaku:

| Analogi | Istilah Python |
|---|---|
| Input | **Parameter** |
| Proses | **Isi Fungsi** |
| Output | **Return / Hasil** |

Mari kita bedah anatomi sebuah fungsi:

```python
def nama_fungsi(parameter):   # 1. def = mendefinisikan, diikuti nama & parameter
    # 2. isi/badan fungsi -- kode yang dijalankan
    hasil = ...
    return hasil               # 3. return -- mengirim hasil keluar (opsional)
```

## 3️⃣ Mendefinisikan & Memanggil Fungsi

Ada **dua langkah** dalam menggunakan fungsi:

1. **Definisikan** — buat "cetak biru"-nya dengan kata kunci `def`
2. **Panggil** — jalankan fungsi dengan menuliskan namanya + `()`

⚠️ **Ingat:** Fungsi yang hanya didefinisikan **TIDAK** akan berjalan sebelum dipanggil!


````python
# 1. Definisikan fungsi
def sapa_pagi():
    print("Selamat pagi, semua!")
    print("Semoga harimu produktif")

# Perhatikan: belum ada output apa pun di sini,
# karena fungsi baru didefinisikan, belum dipanggil.
````

````python
# 2. Panggil fungsi
sapa_pagi()   # baru sekarang isi fungsi dijalankan
````

### 🔄 Alur Eksekusi: "Lompat & Kembali"

Saat fungsi dipanggil, Python akan **meloncat** ke dalam fungsi, menjalankan isinya, lalu **kembali** ke baris setelah pemanggilan. Perhatikan urutan output di bawah ini:


````python
def sapa_pagi():          # (2) definisi fungsi
    print("Pagi!")         # (3) isi fungsi dijalankan saat dipanggil

print("Mulai")              # (1) baris utama jalan duluan
sapa_pagi()                  # -> lompat masuk fungsi, jalankan, lalu kembali
print("Selesai")            # (4) lanjut setelah fungsi selesai
````

**Urutan yang terjadi:**
1. `print("Mulai")` dijalankan
2. Python bertemu `sapa_pagi()` → lompat masuk ke dalam fungsi
3. Isi fungsi dijalankan: `print("Pagi!")`
4. Fungsi selesai → Python kembali & lanjut ke `print("Selesai")`

## 5️⃣ Return — Mengembalikan Hasil

`return` mengirim hasil fungsi **KELUAR** agar bisa disimpan atau dipakai lagi. Ini **berbeda** dengan `print()` yang cuma **menampilkan** ke layar.

**Alur:** Parameter masuk → Diproses di dalam fungsi → `return` hasil keluar


````python
# ✅ Fungsi yang PAKAI return
def tambah(a, b):
    hasil = a + b
    return hasil

x = tambah(3, 4)
print(x)       # 7 -- bisa disimpan & dipakai lagi!
print(x * 2)   # 14 -- karena x benar-benar berisi angka
````

````python
# ❌ Fungsi yang CUMA print(), tanpa return
def tambah_tanpa_return(a, b):
    print(a + b)   # hanya tampil ke layar, tidak dikembalikan

y = tambah_tanpa_return(3, 4)  # "7" tercetak duluan
print(y)   # None -- karena tidak ada return, nilainya hilang!
````

````python
def tambah_tanpa_return(a, b):
    print(a + b)
    return None
````

````python
def tambah(a, b):
    return a + b
````

````python
def tambah_kurang(a, b):
    tambah = a + b
    if a < b:
      return tambah
    kurang = a - b
    return tambah, kurang
````

````python
x = tambah(7, 6)
print(x)
````

````python
hasil = tambah_kurang(a=10, b=7)
print(hasil)
````

````python
hasil = tambah_kurang(a=5, b=7)
print(hasil)
````

## 6️⃣ Parameter vs Return — Perbandingan

| | **Parameter** 📥 | **Return** 📤 |
|---|---|---|
| Arah data | Masuk ke fungsi | Keluar dari fungsi |
| Ditulis di | Dalam tanda kurung `def nama_fungsi(...)` | Dalam isi/badan fungsi |
| Jumlah | Bisa lebih dari satu, dipisah koma | Biasanya satu nilai (bisa juga beberapa sekaligus) |
| Sumber nilai | Argument saat fungsi dipanggil | Hasil pemrosesan di dalam fungsi |
| Bisa disimpan? | N/A (nilai sudah ada saat masuk) | Ya, ke variabel di luar fungsi |


````python
# Contoh yang menggabungkan keduanya:
def luas_lingkaran(jari_jari):     # <- parameter (input)
    pi = 3.14159
    hasil = pi * jari_jari ** 2
    return hasil                    # <- return (output)

luas = luas_lingkaran(7)
print(f"Luas lingkaran = {luas}")
````

## 7️⃣ Global vs Local Scope

Bayangkan dua ruangan:
- 🌐 **Ruang Global** — ruang besar yang bisa dilihat/diakses dari mana saja
- 🔒 **Ruang Local** — ruang kecil milik masing-masing fungsi, hanya bisa diakses dari **dalam** fungsi itu sendiri


````python
nama_kelas = "Python Pemula"   # variabel GLOBAL -- bisa dibaca dari mana saja

def fungsi_a():
    x = 10                      # variabel LOCAL milik fungsi_a
    print("Di dalam fungsi_a, x =", x)
    print("Variabel global tetap bisa diakses:", nama_kelas)

fungsi_a()
````

````python
x
````

````python
# Coba akses variabel lokal DARI LUAR fungsi
def fungsi_b():
    variabel_y = 99
    print("Di dalam fungsi_b, y =", variabel_y)
    return variabel_y

variabel_y = fungsi_b()

# Baris di bawah ini akan ERROR jika dijalankan, karena y hanya "hidup"
# di dalam fungsi_b(). Coba hapus tanda pagar (#) di baris berikut untuk membuktikannya:

# print(y)   # NameError: name 'y' is not defined
````

````python
variabel_y
````

````python
def fungsi_b():
    variabel_y = 99 #variabel local
    print("Di dalam fungsi_b, y =", variabel_y)
    def fungsi_c(variabel_y): #fungsi local
      return variabel_y * 2
    return fungsi_c(variabel_y)
````

````python
fungsi_b()
````

> 🔎 **Kesimpulan:** Variabel yang dibuat **di dalam** fungsi (local) tidak bisa diakses dari luar fungsi. Variabel yang dibuat **di luar** semua fungsi (global) bisa dibaca dari mana saja, termasuk dari dalam fungsi.

## 8️⃣ ⚠️ Jebakan Umum: Scope Error

Dua kesalahan paling sering dilakukan pemula terkait fungsi:

### ❌ Kesalahan 1: Mengakses variabel lokal di luar fungsi


````python
def hitung():
    total = 100
    print("Di dalam fungsi:", total)

hitung()

try:
    print(total)   # akan gagal karena 'total' tidak ada di luar fungsi
except NameError as e:
    print("Error:", e)
````

**✅ Perbaikan:** Gunakan `return` agar nilainya bisa disimpan ke variabel di luar.


````python
def hitung_diperbaiki():
    total = 100
    return total   # kirim keluar

hasil = hitung_diperbaiki()   # simpan hasilnya ke variabel di luar
print("Sekarang bisa diakses:", hasil)
````

### ❌ Kesalahan 2: Lupa menulis `return`


````python
def kali(a, b):
    hasil = a * b   # <- dihitung, tapi TIDAK di-return

x = kali(4, 5)
print(x)   # None, bukan 20!
````

**✅ Perbaikan:** Tambahkan `return hasil` di baris terakhir fungsi.


````python
def kali_diperbaiki(a, b):
    hasil = a * b
    return hasil    # sekarang nilainya bisa dipakai

x = kali_diperbaiki(4, 5)
print(x)   # 20 ✅
````

## 9️⃣ Fungsi = Kode Modular ala Lego 🧱

Setiap fungsi adalah satu **balok Lego**: kecil, punya tugas yang jelas, dan bisa disusun ulang menjadi program yang lebih besar.

Manfaat menulis kode secara modular (dipecah jadi fungsi-fungsi kecil):

| Manfaat | Penjelasan |
|---|---|
| 🧹 **Rapi** | Kode tidak menumpuk & lebih mudah dibaca sekilas |
| 🧪 **Mudah diuji** | Tiap fungsi bisa dicek satu per satu |
| ♻️ **Reusable** | Ditulis sekali, dipakai di banyak tempat |

Contoh: memecah program menjadi beberapa fungsi kecil yang saling bekerja sama.


````python
def input_nilai():
    # Di notebook asli, ini bisa memakai input() dari pengguna.
    # Untuk latihan kali ini kita pakai data contoh:
    return [80, 90, 75, 88, 95]

def hitung_rata_rata(daftar_nilai):
    return sum(daftar_nilai) / len(daftar_nilai)

def tampilkan_hasil(rata_rata):
    print(f"Rata-rata nilai kelas: {rata_rata:.2f}")

# Program utama -- menyusun beberapa fungsi kecil (seperti menyusun Lego!)
nilai = input_nilai()
rata2 = hitung_rata_rata(nilai)
tampilkan_hasil(rata2)
````

## 🎁 Bonus: Sekilas Intip — Apa Itu Class?

Sejauh ini kita sudah jago membungkus **kode** ke dalam fungsi. Nah, ada satu tingkat lebih lanjut yang akan kita pelajari di sesi-sesi mendatang: **Class**.

> **Fungsi** membungkus **kode** agar bisa dipakai ulang.
> **Class** membungkus **kode + data** sekaligus, jadi satu "cetak biru" yang lebih lengkap.

### 🏠 Analogi Class vs Object

| Istilah | Analogi | Penjelasan |
|---|---|---|
| **Class** | 📐 Denah rumah (blueprint) | Rancangan/cetak biru — belum jadi rumah sungguhan |
| **Object** | 🏠 Rumah yang sudah dibangun | Hasil nyata dari class (disebut **instance**). Dari 1 denah, bisa dibangun banyak rumah berbeda! |

💡 **Ingat fungsi?** Di dalam class, fungsi disebut **method** — cara kerjanya sama persis seperti fungsi biasa yang sudah kita pelajari, hanya saja ia "menempel" pada class.

Mari lihat contoh sederhananya:


````python
class Mobil:
    def __init__(self, warna):        # method khusus, otomatis jalan saat objek dibuat
        self.warna = warna              # data milik object (disebut "atribut")

    def klakson(self):                  # method biasa -- mirip fungsi!
        print(f"Mobil {self.warna}: Tin tin!")


# Membuat OBJECT (instance) dari CLASS Mobil
mobil1 = Mobil("Merah")
mobil2 = Mobil("Biru")

mobil1.klakson()   # Mobil Merah: Tin tin!
mobil2.klakson()   # Mobil Biru: Tin tin!
````

Perhatikan: dari **satu class** `Mobil`, kita bisa membuat **banyak object** (`mobil1`, `mobil2`) yang masing-masing punya datanya sendiri (`warna` yang berbeda), tapi berbagi method yang sama (`klakson`). Ini baru **sekilas intip** — materi lengkapnya akan dibahas tuntas nanti! 😉

## 🧩 Bonus: 4 Pilar Object-Oriented Programming (OOP)

Setelah mengenal class, ada **4 prinsip utama** dalam OOP yang akan sering kamu temui ke depannya:

| Pilar | Istilah Indonesia | Penjelasan Singkat |
|---|---|---|
| 🔒 **Encapsulation** | Enkapsulasi | Membungkus data & fungsi dalam satu class, menyembunyikan detail internal dari luar |
| 🌳 **Inheritance** | Pewarisan | Class baru bisa **mewarisi** sifat & perilaku dari class lain, tanpa menulis ulang dari nol |
| 🎭 **Polymorphism** | Polimorfisme | Method dengan **nama sama** bisa berperilaku **berbeda** tergantung objek yang memanggilnya |
| 🧩 **Abstraction** | Abstraksi | Menyembunyikan detail rumit, pengguna cukup tahu **cara pakainya** saja |

Contoh singkat **Inheritance** (pewarisan) agar terbayang sedikit gambarannya:


````python
class Kendaraan:                        # class "induk"
    def __init__(self, merk):
        self.merk = merk

    def info(self):
        print(f"Ini kendaraan merk {self.merk}")


class Motor(Kendaraan):                  # Motor MEWARISI dari Kendaraan
    def klakson(self):
        print("Motor: Ngeeeng!")


motor1 = Motor("Yamaha")
motor1.info()      # method dari class Kendaraan (diwariskan)
motor1.klakson()   # method milik class Motor sendiri
````

✨ **Materi lengkap Class & OOP (termasuk contoh Encapsulation, Polymorphism, dan Abstraction) akan kita bahas tuntas di sesi-sesi mendatang!** Untuk sekarang, cukup pahami gambaran besarnya dulu ya.

### Soal 3 — Fungsi Cek Palindrom
Buat fungsi bernama `cek_palindrom(kata)` yang menerima sebuah **string**, dan **mengembalikan** `True` jika kata tersebut adalah **palindrom** (dibaca sama dari depan maupun belakang), atau `False` jika bukan.

> 💡 Petunjuk: sebuah string bisa dibalik dengan slicing `kata[::-1]`.


````python
# TODO Soal 3: Fungsi Cek Palindrom
def cek_palindrom(kata):
    return kata == kata[::-1]


# --- Uji coba ---
print(cek_palindrom("katak"))   # harapan: True
print(cek_palindrom("python"))  # harapan: False
````
